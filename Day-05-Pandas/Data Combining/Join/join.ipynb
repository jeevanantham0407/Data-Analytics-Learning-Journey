Read The Two Datasets


```python
import pandas as pd

customers = pd.read_csv(r"C:\Users\jeeva\OneDrive\Desktop\Data_Analytics_Learning_Journey\Day-04-Pandas\Data Combining\Join\Dataset\customers_join.csv")
sales = pd.read_csv(r"C:\Users\jeeva\OneDrive\Desktop\Data_Analytics_Learning_Journey\Day-04-Pandas\Data Combining\Join\Dataset\sales_join.csv")

print("Customers Dataset:")
print(customers)

print("\nSales Dataset:")
print(sales)
```

Indexing Setting

```python
customers = customers.set_index("CustomerID")
sales = sales.set_index("CustomerID")
```

Left Join


```python
left_join = customers.join(sales, how="left")

print(left_join)
```

Inner Join

```python
right_join = customers.join(sales, how="inner")

print(right_join)
```

Right Join

```python
right_join = customers.join(sales, how="right")

print(right_join)
```

Outer Join

```python
outer_join = customers.join(sales, how="outer")

print(outer_join)
```
