Read Both Files

```python
import pandas as pd

customers = pd.read_csv(r"C:\Users\jeeva\OneDrive\Desktop\Data_Analytics_Learning_Journey\Day-04-Pandas\Data Combining\Merge\Dataset\customers.csv")
orders = pd.read_csv(r"C:\Users\jeeva\OneDrive\Desktop\Data_Analytics_Learning_Journey\Day-04-Pandas\Data Combining\Merge\Dataset\orders.csv")

print("Customers Dataset:")
print(customers)

print("\nOrders Dataset:")
print(orders)
```

Basic Merge

```python
merged_data = pd.merge(customers, orders, on="CustomerID")

print(merged_data)
```

Inner Merge

```python
inner_merge = pd.merge(customers, orders, on="CustomerID", how="inner")

print(inner_merge)
```

Left Merge

```python
left_merge = pd.merge(customers, orders, on="CustomerID", how="left")

print(left_merge)
```

Right Merge

```python
right_merge = pd.merge(customers, orders, on="CustomerID", how="right")

print(right_merge)
```

Outer Merge

```python
outer_merge = pd.merge(
    customers,
    orders,
    on="CustomerID",
    how="outer"
)

print(outer_merge)
```
