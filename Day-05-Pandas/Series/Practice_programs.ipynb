PROGRAM 1

```python
import pandas as pd

Marks= pd.Series([10,20,30,40,50])

print("Marks:")
print(Marks)

print("\nTotal Marks:")
print(Marks.sum())

print("\nAverage Marks:")
print(Marks.mean())

print("\nMaximim Marks:")
print(Marks.max())

print("\nMinimum Marks:")
print(Marks.min())
```

```python
import pandas as pd

marks=pd.Series({
    "Arun":85,
    "Priya":92,
    "Rahul":78,
    "Divya":95
})

print("Arun's mark:",marks["Arun"])
print("Divya's mark:",marks["Divya"])
print("Average mark:",marks.mean())
print("Highest mark:",marks.max())
```
