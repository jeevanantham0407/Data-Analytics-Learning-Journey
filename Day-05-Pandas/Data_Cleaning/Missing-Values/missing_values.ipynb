Part 1B: Handling Missing Values

Step 1: dropna() — Remove Missing Values

```python
df_drop = df.dropna()

print(df_drop)
```

Step 2: Fill Missing Age with Mean

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())

print(df["Age"])

print(df.isnull().sum())
```

Step 3: Fill Missing Sales with Median

```python
df["Sales"] = df["Sales"].fillna(df["Sales"].median())

print(df["Sales"])

print(df.isnull().sum())
```

Step 4: Fill Missing Categorical Values with Mode

```python
df["Category"] = df["Category"].fillna(df["Category"].mode()[0])

print("Filling Missing Values in Category:")
print(df["Category"])

df["Product"] = df["Product"].fillna(df["Product"].mode()[0])

print("\nFilling Missing Values in Product:")
print(df["Product"])

df["Region"] = df["Region"].fillna(df["Region"].mode()[0])

print("\nFilling Missing Values in Region:")
print(df["Region"])

print(df.isnull().sum())
```

Forward Fill — ffill() And Backward Fill — bfill()

```python
import pandas as pd

data = {
    "OrderID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008],
    
    "CustomerName": [
        "Arun", "Priya", "Rahul", "Divya",
        "Karthik", "Sneha", "Vijay", "Anitha"
    ],
    
    "Age": [25, None, 28, 23, 26, None, 29, 24],
    
    "Product": [
        "Laptop", "Mouse", "Keyboard", None,
        "Monitor", "Laptop", "Mouse", "Keyboard"
    ],
    
    "Category": [
        "Electronics", "Accessories", "Accessories", "Electronics",
        "Electronics", "Electronics", None, "Accessories"
    ],
    
    "Region": [
        "Chennai", "Chennai", None, "Bangalore",
        "Chennai", "Bangalore", "Chennai", None
    ],
    
    "Sales": [55000, 1200, 2500, 45000, None, 60000, 1500, 3000]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

df_ffill = df.ffill()

print("\nForward Fill:")
print(df_ffill)

# Backward Fill
df_bfill = df.bfill()

print("\nBackward Fill:")
print(df_bfill)
```
