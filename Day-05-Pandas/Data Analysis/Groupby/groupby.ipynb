Read the Dataset

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\jeeva\OneDrive\Desktop\Data_Analytics_Learning_Journey\Day-04-Pandas\Data Analysis\Groupby\Dataset\sales_groupby.csv")

print("Dataset:")
print(df)
```

Group by Region

```python
region_sales = df.groupby("Region")["Sales"].sum()

print(region_sales)
```

Average Sales by Region 

```python
avg_region_sales = df.groupby("Region")["Sales"].mean()

print(avg_region_sales)
```

Number of Orders by Region

```python
region_orders_count = df.groupby("Region")["OrderID"].count()

print(region_orders_count)
```

Minimum Sales by Region

```python
min_region_sales = df.groupby("Region")["Sales"].min()

print(min_region_sales)
```

Maximum Sales by Region

```python
max_region_sales = df.groupby("Region")["Sales"].max()

print(max_region_sales)
```

Group by Category

```python
print("Sales by Category:")
print(df.groupby("Category")["Sales"].sum())

print("\nSales by Product:")
print(df.groupby("Product")["Sales"].sum())

print("\nSales by Salesperson:")
print(df.groupby("Salesperson")["Sales"].sum())
```

Multiple Calculations

```python
result = df.groupby("Region")["Sales"].agg(
    ["sum", "mean", "min", "max"]
)

print(result)
```

Group by Multiple Columns


```python
print("Multiple Aggregations:")

print(
    df.groupby(["Region", "Category"])["Sales"].sum()
)
```
