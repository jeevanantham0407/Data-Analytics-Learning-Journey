Read the csv file


```python
import pandas as pd

df = pd.read_csv(r"C:\Users\jeeva\OneDrive\Desktop\Data_Analytics_Learning_Journey\Day-04-Pandas\Data Analysis\Groupby\Dataset\sales_groupby.csv")


print("DATASET:")
print(df)
```

Basic Pivot Table


```python
basic_pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Region",
    aggfunc="sum"
)

print(basic_pivot)
```

Sales by Category

```python
pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Category",
    aggfunc="sum"
)

print(pivot)
```

Region and Category Report

```python
pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Region",
    columns="Category",
    aggfunc="sum"
)

print(pivot)
```

Average Sales

```python
pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Product",
    aggfunc="mean"
)

print(pivot)
```

Multiple Calculations

```python
pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Region",
    aggfunc=["sum", "mean", "min", "max"]
)

print(pivot)
```
