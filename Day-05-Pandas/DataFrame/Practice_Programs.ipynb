PROGRAM 1

```python
import pandas as pd

data={
    "Name": ["arjun", "jeeva", "kishore", "jon"],
    "Age": [22, 23, 22, 23],
    "Marks": [50, 60, 70, 80],
    "Department": ["IT", "CSE", "AI&DS", "ECE"]
}

df=pd.DataFrame(data)

print("DataFrame:")
print(df)

print("\nShape")
print(df.shape)

print("\nColumns")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

```

PROGRAM 2

```python
import pandas as pd

data={
    "Name": ["Arun", "Priya", "Rahul", "Divya"],
    "Age": [25, 24, 26, 23],
    "Salary": [30000, 35000, 40000, 32000]
}

df=pd.DataFrame(data)

print("DataFrame:")
print(df)

print("\nShape:")
print(df.shape)

print("\nAverage Salary:")
print(df["Salary"].mean())

print("\nHighest Salary:")
print(df["Salary"].max())

print("\nLowest Salary:")
print(df["Salary"].min())
```

PROGRAM 4

```python
import pandas as pd

data={
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor"],
    "Price": [50000, 1000, 2000, 15000],
    "Quantity": [2, 5, 3, 4]
}

df=pd.DataFrame(data)

print("DATA:")
print(df)

print("\nDescribe:")
print(df.describe())

print("\nAverage Price:")
print(df["Price"].mean())

print("\nHighest Price:")
print(df["Price"].max())

print("\nLowest Price:")
print(df["Price"].min())

print("\nAverage Quantity:")
print(df["Quantity"].mean())
```

```python
import pandas as pd

data={
    "name": ["Jeeva", "harish", "arun", "kishore", "charu"],
    "age": [22, 23, 22, 24, 25],
    "dept": ["CSE", "AI&DS", "ECE", "IT", "CSE"],
    "CGPA": [8.01, 8.02, 7.09, 9.00, 7.65]
}

df=pd.DataFrame(data)

print("TOP:")
print(df.head())

print("\nBOTTOM:")
print(df.tail())

print("\nSHAPE:")
print(df.shape)

print("\nSIZE:")
print(df.size)

print("\nCOLUMNS:")
print(df.columns)

print("\nINDEX:")
print(df.index)

print("\nDTYPES:")
print(df.dtypes)

print("\nINFO:")
print(df.info())

print("\nDESCRIBE:")
print(df.describe())
```
