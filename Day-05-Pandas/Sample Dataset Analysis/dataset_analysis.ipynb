Load the dataset

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\jeeva\OneDrive\Desktop\Data_Analytics_Learning_Journey\Day-04-Pandas\Sample Dataset Analysis\Dataset\students.csv")

print(df)
```

Understand the dataset

```python
print("Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nFirst 5 Rows:")
print(df.head())

print("\nMissing Values:")
print(df.isnull().sum())
```

Number of students

```python
print("Number of Students:", df.shape[0])
```

Average marks

```python
average_marks = df["Marks"].mean()

print("Average Marks:", average_marks)
```

Highest marks

```python
highest_mark = df["Marks"].max()

print("Highest Mark:", highest_mark)
```

Lowest marks

```python
lowest_mark = df["Marks"].min()

print("lowest mark:", lowest_mark)
```

Find the student with the highest marks

```python
top_student = df.loc[df["Marks"].idxmax()]

print(top_student)
```

Find the lowest-scoring student

```python
lowest_student = df.loc[df["Marks"].idxmin()]

print(lowest_student)
```

Students scoring above 85

```python
high_scorers = df.loc[df["Marks"] > 85,["Marks", "Name"]]

print(high_scorers)
```

Top 3 students

```python
top_3 = df.sort_values(
    "Marks",
    ascending = False
).head(3)

print(top_3)
```

Number of students in each department

```python
department_count = df["Department"].value_counts()

print(department_count)
```

Average marks by department

```python
department_average = df.groupby("Department")["Marks"].mean()

print(department_average)
```

Highest average department

```python
department_average = df.groupby("Department")["Marks"].mean()

print(department_average.idxmax())
```
