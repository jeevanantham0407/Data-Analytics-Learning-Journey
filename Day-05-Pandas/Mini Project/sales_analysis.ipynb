Load sales dataset

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\jeeva\OneDrive\Desktop\Data_Analytics_Learning_Journey\Day-04-Pandas\Mini Project\Dataset\sales.csv")

print("Dataset:")
print(df)
```

Dataset Inspection

```python
print("\nFIRST 5 ROWS:")
print(df.head())

print("\nDATASET SHAPE:")
print(df.shape)

print("\nCOLUMNS:")
print(df.columns)

print("\nDATA TYPES:")
print(df.dtypes)

print("\nMISSING VALUES:")
print(df.isnull().sum())
```

Calculate Revenue

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
print(df)
```

Check the result

```python
print(df[["Product", "Quantity", "UnitPrice", "Revenue"]])
```

Find Total Revenue

```python
total_revenue = df["Revenue"].sum()

print("Total Revenue:", total_revenue)
```

Find Total Quantity Sold

```python
total_quantity = df["Quantity"].sum()

print("Total Quantity Sold:", total_quantity)
```

Find Average Order Revenue

```python
average_order_revenue = df["Revenue"].mean()

print("Average Order Revenue:", average_order_revenue)
```

Find the Highest-Value Order

```python
highest_order = df.loc[df["Revenue"].idxmax()]

print("\nHighest Revenue Order:")
print(highest_order)
```

Product-wise Revenue

```python
product_revenue = df.groupby("Product")["Revenue"].sum()

print(product_revenue)
```

Sort Product Revenue

```python
product_revenue = (
    df.groupby("Product")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print(product_revenue)
```

Region-wise Revenue

```python
region_revenue = (
    df.groupby("Region")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print(region_revenue)
```

Salesperson Performance

```python
salesperson_revenue = (
    df.groupby("Salesperson")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print(salesperson_revenue)
```

Category Revenue

```python
category_revenue = (
    df.groupby("Category")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print(category_revenue)
```

Business Insights

```python
# --------------------------------------------------
# Identify Best Product
# --------------------------------------------------

best_product = product_revenue.idxmax()
best_product_revenue = product_revenue.max()

print("\nBEST-PERFORMING PRODUCT:")
print("Product:", best_product)
print("Revenue:", best_product_revenue)


# --------------------------------------------------
# Identify Best Region
# --------------------------------------------------

best_region = region_revenue.idxmax()
best_region_revenue = region_revenue.max()

print("\nBEST-PERFORMING REGION:")
print("Region:", best_region)
print("Revenue:", best_region_revenue)


# --------------------------------------------------
# Identify Best Category
# --------------------------------------------------

best_category = category_revenue.idxmax()
best_category_revenue = category_revenue.max()

print("\nBEST-PERFORMING CATEGORY:")
print("Category:", best_category)
print("Revenue:", best_category_revenue)


# --------------------------------------------------
# Identify Best Salesperson
# --------------------------------------------------

best_salesperson = salesperson_revenue.idxmax()
best_salesperson_revenue = salesperson_revenue.max()

print("\nBEST-PERFORMING SALESPERSON:")
print("Salesperson:", best_salesperson)
print("Revenue:", best_salesperson_revenue)
```
