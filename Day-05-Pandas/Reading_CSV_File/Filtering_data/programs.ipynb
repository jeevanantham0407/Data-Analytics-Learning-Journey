```python
import pandas as pd


df=pd.read_csv(r"C:\Users\jeeva\OneDrive\Desktop\Data_Analytics_Learning_Journey\Day-04-Pandas\Reading_CSV_File\Dataset\student.csv")

# Students with marks greater than 85
print("MARKS GREATER THAN 85:")
print(df[df["Marks"] > 85])

# Students with marks 85 or above
print("\nMARKS 85 OR ABOVE:")
print(df[df["Marks"] >= 85])

# Students with marks below 85
print("\nMARKS BELOW 85:")
print(df[df["Marks"] < 85])

# CSE students
print("\nCSE STUDENTS:")
print(df[df["Department"] == "CSE"])

# Students not from CSE
print("\nNON-CSE STUDENTS:")
print(df[df["Department"] != "CSE"])

# Students aged 21 or above
print("\nSTUDENTS AGED 21 OR ABOVE:")
print(df[df["Age"] >= 21])

# Marks greater than 85 AND age 21 or above
print("\nMARKS > 85 AND AGE >= 21:")
print(df[(df["Marks"] > 85) & (df["Age"] >= 21)])

# Students from CSE OR IT
print("\nCSE OR IT STUDENTS:")
print(df[(df["Department"] == "CSE") | (df["Department"] == "IT")])

# Marks between 80 and 90
print("\nMARKS BETWEEN 80 AND 90:")
print(df[df["Marks"].between(80, 90)])

# Students from CSE or ECE using isin
print("\nCSE OR ECE STUDENTS:")
print(df[df["Department"].isin(["CSE", "ECE"])])

# Select Name and Marks for students scoring above 85
print("\nNAME AND MARKS FOR STUDENTS ABOVE 85:")
print(df.loc[df["Marks"] > 85, ["Name", "Marks"]])
```
