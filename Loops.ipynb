1.Multiplication Table

```python
number = 6

for i in range(1, 11):
    print(f"{i} * {number} = {number * i}")
```

2.Sum of Even Numbers

```python
total = 0

for i in range(1, 101):
    if i % 2 == 0:
        total += i

print("Sum:", total)
```

3.Count Positive and Negative Numbers

```python
numbers = [10, -5, 20, -8, 15, -2]

positive = 0
negative = 0

for num in numbers:
    if num > 0:
        positive += 1
    else:
        negative += 1

print("Positive:", positive)
print("Negative:", negative)
```

4.Reverse Counting

```python
for i in range(10, 0, -1):
    print(i)

print("Blast Off!")
```

5.Password Verification

```python
correct_password = "python123"

while True:
    password = input("Enter Password: ")

    if password == correct_password:
        print("Login Successful")
        break

    print("Incorrect Password")
```

6.ATM PIN Verification (3 Attempts)

```python
correct_pin = 1234
attempts = 0

while attempts < 3:
    pin = int(input("Enter PIN: "))

    if pin == correct_pin:
        print("Access Granted")
        break

    attempts += 1
    print("Incorrect PIN")

if attempts == 3:
    print("Card Blocked")
```

7.Skip Multiples of 5

```python
for i in range(1, 31):

    if i % 5 == 0:
        continue

    print(i)
```

8.Find the First Number Divisible by 17

```python
for i in range(50, 101):

    if i % 17 == 0:
        print("First Number:", i)
        break
```

9.Guess the Secret Number

```python
secret = 8

while True:

    guess = int(input("Guess the Number: "))

    if guess < 1:
        print("Enter a Positive Number")
        continue

    if guess == secret:
        print("Correct!")
        break

    print("Wrong Guess")
```

10.Student Result Processing

```python
marks = [85, 42, 90, 35, 76, 48]

for mark in marks:

    if mark < 50:
        continue

    print("Passed:", mark)
```
