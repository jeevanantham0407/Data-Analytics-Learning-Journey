Grade Calculator

```python


print("===== Student Grade Calculator =====")

student_name = input("Enter Student Name: ")

python = int(input("Enter Python Marks: "))
sql = int(input("Enter SQL Marks: "))
excel = int(input("Enter Excel Marks: "))
power_bi = int(input("Enter Power BI Marks: "))
statistics = int(input("Enter Statistics Marks: "))


total = python + sql + excel + power_bi + statistics
average = total / 5


if python >= 35 and sql >= 35 and excel >= 35 and power_bi >= 35 and statistics >= 35:
    result = "PASS"

    if average >= 90:
        grade = "A+"
    elif average >= 80:
        grade = "A"
    elif average >= 70:
        grade = "B"
    elif average >= 60:
        grade = "C"
    elif average >= 50:
        grade = "D"
    else:
        grade = "E"
else:
    result = "FAIL"
    grade = "F"

print("\n===== RESULT =====")
print("Student Name :", student_name)
print("Total Marks  :", total)
print("Average      :", round(average, 2))
print("Result       :", result)
print("Grade        :", grade)
```
