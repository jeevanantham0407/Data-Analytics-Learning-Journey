1.Arithmetic Operations

```python
a = 20
b = 6

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)
```

2.Compare Two Numbers

```python
a = 15
b = 20

print("Equal:", a == b)
print("Not Equal:", a != b)
print("Greater Than:", a > b)
print("Less Than:", a < b)
print("Greater Than or Equal:", a >= b)
print("Less Than or Equal:", a <= b)
```

3.Update Salary Using Assignment Operators

```python
salary = 25000

salary += 5000

print("Updated Salary:", salary)
```

4.Apply Multiple Assignment Operators

```python
num = 20

num += 10
print("After += :", num)

num -= 5
print("After -= :", num)

num *= 2
print("After *= :", num)

num //= 5
print("After //= :", num)
```

5.Check Voting Eligibility Using Logical Operators

```python
age = 21
citizen = True

print(age >= 18 and citizen)
```

6.Check Scholarship Eligibility

```python
marks = 88
sports_quota = True

print(marks > 90 or sports_quota)
```

7.Check Membership in a List

```python
courses = ["Python", "SQL", "Excel", "Power BI"]

print("Python" in courses)
print("Java" not in courses)
```

8.Check a Character in a String

```python
word = "Data Analytics"

print("a" in word)
print("z" not in word)
```

9.Compare Identity of Two Lists

```python
list1 = [1, 2, 3]
list2 = list1
list3 = [1, 2, 3]

print(list1 is list2)
print(list1 is list3)
print(list1 == list3)
```

10.Employee Bonus Eligibility

```python
experience = 6
rating = 4.5

eligible = experience >= 5 and rating > 4

print("Bonus Eligible:", eligible)
```
