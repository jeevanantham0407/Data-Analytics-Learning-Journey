Query parameters

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

params = {
    "userId": 1
}

response = requests.get(url, params=params)

posts = response.json()

print("Status Code:", response.status_code)
print("Number of Posts:", len(posts))

print("\n")
for post in posts:
    print(post["title"])
```
