Check the status code

```python
import requests

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

print("Status Code:",response.status_code)
```

Get the actual JSON

```python
import requests

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

data = response.json()

for user in data:
    print(user)
    print()
```

Check the data type

```python
print(type(data))
```

Get the first user

```python
print("First User:")
print(data[0])
print("\nFirst user's mail:")
print(data[0]["email"])

print("\nSecond Data:")
print(data[1])
print("\nSecond User's Name:")
print(data[2]["name"])
```

Get all user names

```python
import requests

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

users = response.json()

print("User Names:")

for i, user in enumerate(users, start=1):
    print(i, ".", user["name"])
    
```

Get name + email

```python
import requests

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

users = response.json()

for user in users:
    print("Name:", user["name"])
    print("Email:", user["email"])
    print()
```
