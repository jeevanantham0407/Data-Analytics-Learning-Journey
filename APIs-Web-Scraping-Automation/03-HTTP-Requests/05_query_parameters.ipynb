Basic query parameter

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

params = {
    "userId": 1
}

response = requests.get(url, params=params)

print("Status Code:", response.status_code)

posts = response.json()

for post in posts:
    print("Post ID:", post["id"])
    print("Title:", post["title"])
    print()
```

Multiple parameters

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

params = {
    "userId": 1,
    "_limit": 3
}

response = requests.get(url, params=params)

print("Status Code:", response.status_code)

posts = response.json()

for post in posts:
    print("Post ID:", post["id"])
    print("Title:", post["title"])
```

See the actual URL

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

params = {
    "userId": 1,
    "_limit": 3
}

response = requests.get(url, params=params)

print("URL:", response.url)
print("Status Code:", response.status_code)

posts = response.json()

for post in posts:
    print("Post ID:", post["id"])
    print("Title:", post["title"])
```
