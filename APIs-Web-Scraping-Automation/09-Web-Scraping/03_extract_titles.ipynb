Extract all quote text

```python
import requests
from bs4 import BeautifulSoup

url = "https://quotes.toscrape.com/"

response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

quotes = soup.find_all("span", class_="text")

print("First Quote:")
quote = soup.find("span", class_="text")
print(quote.text)

for i, quote in enumerate(quotes, start=1):
    print("\nAll Quotes:")
    print(f"{i}. {quote.text}")
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://quotes.toscrape.com/"
response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

link = soup.find("a", href="/author/Albert-Einstein")

print(link.get("href"))
```

Extract authors

```python
authors = soup.find_all("small", class_="author")

for i, author in enumerate(authors, start=1):
    print(f"{i}. {author.text}")
```

Extract quotes and authors together

```python
quotes = soup.find_all("span", class_="text")
authors = soup.find_all("small", class_="author")

for i, (quote, author) in enumerate(zip(quotes, authors), start=1):
    print("Quote", i, ":", quote.text)
    print("Author:", author.text)
    print("-" * 50)
```

Extract tags

```python
tags = soup.find_all("a", class_="tag")

for i, tag in enumerate(tags, start=1):
    print(f"{i}. {tag.text}")
```

Extract links

```python
import requests
from bs4 import BeautifulSoup

url = "https://quotes.toscrape.com/"

response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

links = soup.find_all("a")

for i, link in enumerate(links, start=1):
    print(f"{i}. Link Text: {link.text.strip()}")
    print("URL:", link.get("href"))
    print("-" * 30)
```

Extract only author links

```python
author_links = soup.select("small.author + a")

for link in author_links:
    print("Author Link Text:", link.text.strip())
    print("URL:", link.get("href"))
    print("-" * 30)
```
