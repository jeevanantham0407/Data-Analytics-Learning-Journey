Use BeautifulSoup

```python
import requests
from bs4 import BeautifulSoup

url = "https://example.com"

response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

print(soup.prettify())
```

Extract the page title

```python
title = soup.title

print("Title:")
print(title)

print("\nTitle As Text:")
print(title.text)
```

Extract the heading

```python
heading = soup.h1

print("Heading:")
print(heading)

print("\nHeading As Text:")
print(heading.text)

print("\nBody Content:")
print(soup.body.text)
```
