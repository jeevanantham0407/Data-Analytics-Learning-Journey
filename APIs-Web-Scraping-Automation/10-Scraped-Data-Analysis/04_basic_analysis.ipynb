```python
import pandas as pd

df = pd.read_csv(r"C:\Users\jeeva\OneDrive\Desktop\Data_Analytics_Learning_Journey\APIs-Web-Scraping-Automation\09-Web-Scraping\quotes.csv")

print("Dataset:")
print(df.to_string())

print("\n")

print("Total Quotes:", len(df))

print("\n")

print("\nUnique Authors:")
print(df["author"].nunique())

print("\n")

print("\nAuthor Frequency:")
print(df["author"].value_counts())

print("\n")

print("\nMost Frequent Author:")
print(df["author"].value_counts().idxmax())

print("\n")

print("\nLeast Frequent Author:")
print(df["author"].value_counts().idxmin())

print("\n")

print("\nAlbert Einstein Quotes:")
print(df[df["author"] == "Albert Einstein"])

print("\n")

df["tag_count"] = df["tags"].str.split(", ").str.len()

print("\nTag Count:")
print(df[["author", "tag_count"]])

highest_tags = df.loc[df["tag_count"].idxmax()]

print("\n")

print("\nQuote with Most Tags:")
print(highest_tags)
```
