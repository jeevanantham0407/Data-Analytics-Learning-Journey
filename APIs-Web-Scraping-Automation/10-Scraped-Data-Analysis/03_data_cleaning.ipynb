```python
import pandas as pd

df = pd.read_csv(r"C:\Users\jeeva\OneDrive\Desktop\Data_Analytics_Learning_Journey\APIs-Web-Scraping-Automation\09-Web-Scraping\quotes.csv")

print("Raw Dataset:")
print(df.to_string())

print("\n")

print("Missing values:")
print(df.isnull().sum())

print("\n")

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\n")

df = df.drop_duplicates()

df["quote"] = df["quote"].str.strip()
df["author"] = df["author"].str.strip()
df["tags"] = df["tags"].str.strip()

print("\nCleaned Data:")
print(df.to_string(index=False))
```
