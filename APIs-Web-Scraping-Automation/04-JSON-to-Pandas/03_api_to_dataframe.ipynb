Import libraries

```python
import requests
import pandas as pd
```

Get data from the API

```python
url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

users = response.json()
```

Convert to DataFrame

```python
df = pd.DataFrame(users)

print("Data:")
print("\n", df.to_string())
```

Inspect the Data

```python
print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nFirst 5 rows:")
print(df.head().to_string())
```

Extract Nested Data

```python
df["city"] = df["address"].apply(lambda x: x["city"])
df["company_name"] = df["company"].apply(lambda x: x["name"])

print(df[["name", "email", "city", "company_name"]].to_string())
```

Select Only Useful Columns

```python
final_df = df[
    ["id", "name", "username", "email", "city", "company_name"]
]

print(final_df.to_string())
```
