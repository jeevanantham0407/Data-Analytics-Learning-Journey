sample nested JSON

```python
import json

data = '''
{
    "id": 1,
    "name": "Jeevan",
    "email": "jeevan@example.com",
    "address": {
        "city": "Chennai",
        "zipcode": "600001"
    },
    "company": {
        "name": "ABC Analytics"
    }
}
'''

student = json.loads(data)

print("Name:", student["name"])
print("Email:", student["email"])
print("City:", student["address"]["city"])
print("Zipcode:", student["address"]["zipcode"])
print("Company:", student["company"]["name"])
```

Nested JSON from the real API

```python
import requests

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

users = response.json()

for user in users:
    print("Name:", user["name"])
    print("Email:", user["email"])
    print("City:", user["address"]["city"])
    print("Company:", user["company"]["name"])
    print("-" * 30)
```

Deeper nested JSON

```python
import requests

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

users = response.json()

for user in users:
    print("Name:", user["name"])
    print("Latitude:", user["address"]["geo"]["lat"])
    print("Longitude:", user["address"]["geo"]["lng"])
    print()
```
