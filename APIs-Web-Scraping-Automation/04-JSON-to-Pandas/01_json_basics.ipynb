```python
import json

data = '''
{
    "id": 1,
    "name": "Jeevan",
    "marks": 85,
    "department": "Data Analytics"
}
'''

student = json.loads(data)

for key, value in student.items():
    print(key, ":", value)

print("\n", type(student))

print("\nName:", student["name"])
print("Marks:", student["marks"])
print("Department:", student["department"])
```

JSON List

```python
import json

data = '''
[
    {
        "id": 1,
        "name": "Jeevan",
        "marks": 85
    },
    {
        "id": 2,
        "name": "Arun",
        "marks": 90
    },
    {
        "id": 3,
        "name": "Karthik",
        "marks": 78
    }
]
'''

students = json.loads(data)

for student in students:
    for key, value in student.items():
        print(key, ":", value)

print("\n", type(students))

for student in students:
    print("\nName:", student["name"])
    print("Marks:", student["marks"])
    print()
```
