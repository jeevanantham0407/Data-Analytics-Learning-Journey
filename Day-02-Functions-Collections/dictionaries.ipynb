1.Store and Display Student Marks

```python
students = {
    "John": 82,
    "Mary": 95,
    "Alex": 76,
    "Sara": 89
}

print("Student Marks")

for name in students:
    print(name, ":", students[name])
```

2.Calculate Total and Average Marks

```python
students = {
    "John": 82,
    "Mary": 95,
    "Alex": 76,
    "Sara": 89
}

total = sum(students.values())

average = total / len(students)

print("Total Marks:", total)
print("Average Marks:", average)
```

3.Update Student Marks

```python
students = {
    "John": 82,
    "Mary": 95,
    "Alex": 76
}

students["Alex"] = 88

print(students)
```

4.Delete a Student Record

```python
students = {
    "John": 82,
    "Mary": 95,
    "Alex": 76
}

del students["Mary"]

print(students)
```

5.Display All Keys and Values

```python
student = {
    "Name": "John",
    "Age": 21,
    "Course": "Data Analytics"
}

print("Keys")

for key in student.keys():
    print(key)

print()

print("Values")

for value in student.values():
    print(value)
```
