1.Find the Largest Number in a List

```python
numbers = [45, 78, 12, 99, 34]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print("Largest Number:", largest)
```

2.Find the Smallest Number in a List

```python
numbers=[1,2,3,4,5]

smallest = numbers[0]

for num in numbers:
    if num < smallest:
        smallest = num

print("smallest number:", smallest)
```

3.Calculate the Sum of All Elements

```python
numbers = [10, 20, 30, 40, 50]

total = 0

for num in numbers:
    total += num

print("Total:", total)
```

4.Remove Duplicates from a List

```python
numbers = [10, 20, 30, 20, 40, 10, 50]

unique = []

for num in numbers:
    if num not in unique:
        unique.append(num)

print("Original List:", numbers)
print("Without Duplicates:", unique)
```

5.Reverse a List

```python
numbers = [10, 20, 30, 40, 50]

reversed_list = []

for i in range(len(numbers)-1, -1, -1):
    reversed_list.append(numbers[i])

print("Original:", numbers)
print("Reversed:", reversed_list)
```

6.Sort a List in Ascending and Descending Order

```python
numbers = [45, 12, 78, 34, 20]

ascending = numbers.copy()
ascending.sort()

descending = numbers.copy()
descending.sort(reverse=True)

print("Ascending:", ascending)
print("Descending:", descending)
```

7.Count Even and Odd Numbers

```python
numbers = [12, 17, 20, 25, 30, 41]

even = 0
odd = 0

for num in numbers:
    if num % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even Numbers:", even)
print("Odd Numbers:", odd)
```

8.Merge Two Lists

```python
list1 = [10, 20, 30]
list2 = [40, 50, 60]

merged = list1 + list2

print("Merged List:", merged)
```

9.Find Common Elements in Two Lists


```python
list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 60, 70]

common = []

for num in list1:
    if num in list2:
        common.append(num)

print("Common Elements:", common)
```
