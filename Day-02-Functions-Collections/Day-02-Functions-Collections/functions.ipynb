1.Find the Square of a Number

```python
def number(num):
    return num * num

num=int(input("Enter a number:"))
print("square:",number(num))
```

2.Area of a Rectangle

```python
def area(length, width):
    return length * width

length=float(input("Enter a length:"))
width=float(input("Enter a width:"))

print("Area Of Reactangle:",area(length, width))
```

3.Largest of Three Numbers

```python
def largest(a,b,c):
    if a>b and a>c:
        return a
    elif b>a and b>c:
        return b
    else:
        return c
x=int(input("Enter a value:"))
y=int(input("Enter a value:"))
z=int(input("Enter a value:"))

print("The Largest Value:",largest(x,y,z))
```

4.Check Even or Odd

```python
def check(number):

    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

num = int(input("Enter Number: "))

print(check(num))
```

5.Factorial Using Function

```python
def factorial(n):

    fact = 1

    for i in range(1, n + 1):
        fact *= i

    return fact

number = int(input("Enter Number: "))

print("Factorial:", factorial(number))
```

6.Average of Three Marks

```python
def average(m1, m2, m3):
    return (m1 + m2 + m3) / 3

print("Average:", average(80, 75, 95))
```

7.Positional Arguments

```python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student("Rahul", 21)
```

8.Keyword Arguments

```python
def employee(name, salary):
    print("Name:", name)
    print("Salary:", salary)

employee(salary=45000, name="Anitha")
```

9.Default Arguments

```python
def greet(name="Guest"):
    print("Welcome", name)

greet()
greet("Jeevanantham")
```

10.Local and Global Variables

```python
message = "Global Variable"

def display():
    message = "Local Variable"
    print(message)

display()

print(message)
```

11.Lambda Function

```python
square = lambda x: x ** 2

print(square(9))
```

12.Sort a List Using Lambda

```python
students = [
    ("John", 82),
    ("Mary", 95),
    ("Alex", 76),
    ("Sara", 89)
]

students.sort(key=lambda student: student[1], reverse=True)

for student in students:
    print(student)
```
