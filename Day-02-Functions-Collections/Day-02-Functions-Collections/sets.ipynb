1.Remove Duplicates from a List Using a Set

```python
numbers = [10, 20, 30, 20, 40, 10, 50, 30]

unique_numbers = set(numbers)

print("Original List :", numbers)
print("Unique Values :", unique_numbers)
```

2.Find Common Subjects Between Two Students

```python
student1 = {"Python", "SQL", "Excel", "Power BI"}
student2 = {"SQL", "Python", "Tableau", "Java"}

common_subjects = []

for subject in student1:
    if subject in student2:
        common_subjects.append(subject)

print("Common Subjects:", common_subjects)
```

3.Union of Two Sets

```python
set1 = {10, 20, 30, 40}

set2 = {30, 40, 50, 60}

result = set1.union(set2)

print("Union:", result)
```

4.Difference and Symmetric Difference

```python
A = {1, 2, 3, 4, 5}

B = {4, 5, 6, 7, 8}

print("Difference (A - B):", A.difference(B))

print("Symmetric Difference:", A.symmetric_difference(B))
```

5.Student Club Management Using Set Operations

```python
coding = {"John", "Mary", "Alex"}

robotics = {"Mary", "David", "Alex"}

coding.add("Sara")

robotics.remove("David")

print("Coding Club:", coding)

print("Robotics Club:", robotics)

print("Members in Both Clubs:")

print(coding.intersection(robotics))
```
