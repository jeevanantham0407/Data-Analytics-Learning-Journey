1.Store and Display Student Names

```python
students = ("John", "Mary", "Alex", "Sara", "David")

print("Student Names:")

for student in students:
    print(student)
```

2.Access the First and Last Element

```python
languages = ("Python", "SQL", "Excel", "Power BI", "Tableau")

print("First Element :", languages[0])
print("Last Element  :", languages[-1])
```

3.Tuple Slicing

```python
numbers = (10, 20, 30, 40, 50, 60, 70)

print("Original Tuple:", numbers)

print("First Three Elements:", numbers[:3])

print("Middle Elements:", numbers[2:5])

print("Last Two Elements:", numbers[-2:])
```

4.Tuple Packing and Unpacking

```python
employee = (101, "Rahul", "Data Analyst", 45000)

emp_id, name, job, salary = employee

print("Employee ID :", emp_id)
print("Name        :", name)
print("Job         :", job)
print("Salary      :", salary)
```

5.Count and Find the Index of an Element

```python
numbers = (10, 20, 30, 20, 40, 20, 50)

print("Tuple:", numbers)

print("Count of 20:", numbers.count(20))

print("Index of 40:", numbers.index(40))
```
