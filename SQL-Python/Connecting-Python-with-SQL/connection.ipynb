3. Connect Python to MySQL

```python
import mysql.connector

connection = mysql.connector.connect(
    host="127.0.0.1",
    port=3306,
    user="root",
    password="Jeeva$2004",
    database= "sales_db"
)

if connection.is_connected():
    print("Connected successfully!")
```

4. Create a Cursor

```python
cursor = connection.cursor()
```

Run a SQL Query

```python
cursor.execute("SELECT * FROM sales")
```

Get the Data

```python
rows = cursor.fetchall()

for row in rows:
    print(row)
```

fetchone() - It Return One Row


```python
results = cursor.fetchone()
print(results)
```

fetchall() - It Returns all rows

```python
result = cursor.fetchall()

for rows in result:
    print(rows)
```
