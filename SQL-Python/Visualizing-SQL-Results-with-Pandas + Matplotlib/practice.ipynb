SQL Data → Bar Chart <br>

Import libraries

```python
import mysql.connector
import pandas as pd
import matplotlib.pyplot as plt
```

Connect to MySQL

```python
connection = mysql.connector.connect(
    host="127.0.0.1",
    port=3306,
    user="root",
    password="your_password",
    database= "sales_db"
)

if connection.is_connected():
    print("Connected successfully!")
```

Write the SQL query

```python
query = """

SELECT
    Region,
    SUM(Sales) AS Total_Sales
FROM sales
GROUP BY Region

"""
```

Load SQL result into Pandas

```python
df = pd.read_sql(query, connection)

print(df)
```

Create the bar chart

```python
plt.figure(figsize=(8, 5))

plt.bar(df["Region"], df["Total_Sales"])

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.grid(axis="y")

plt.show()
```

SQL Results → Line Chart

```python
query1 = """
SELECT
    category,
    SUM(sales) AS Total_Sales
FROM sales
GROUP BY category
ORDER BY Total_Sales DESC
"""

df1 = pd.read_sql(query1, connection)

print(df1)

plt.figure(figsize=(8, 5))

plt.plot(
    df1["category"],
    df1["Total_Sales"],
    marker="o"
)

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.grid()

plt.show()
```

SQL Results → Pie Chart

```python
query5 = """
SELECT
    Category,
    SUM(Sales) AS Total_Sales
FROM sales
GROUP BY Category
"""

df5 = pd.read_sql(query5, connection)

print(df)

plt.figure(figsize=(8, 5))

explode = [0.1, 0, 0]

plt.pie(
    df5["Total_Sales"],
    labels=df5["Category"],
    autopct="%1.1f%%",
    explode=explode
)

plt.title("Sales Distribution by Category")

plt.show()
```

Adding Data Labels to SQL Results

```python
query7 = """
SELECT
    Category,
    SUM(Sales) AS Total_Sales
FROM sales
GROUP BY Category
"""

df7 = pd.read_sql(query7, connection)

plt.figure(figsize=(8, 5))

plt.bar(df7["Category"], df7["Total_Sales"])

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.grid(axis="y")

# Add data labels
for i, value in enumerate(df7["Total_Sales"]):
    plt.text(i, value, str(value), ha="center")

plt.tight_layout()

plt.show()
```

Saving SQL Visualizations

```python
query8 = """
SELECT
    Region,
    SUM(Sales) AS Total_Sales
FROM sales
GROUP BY Region
"""

df8 = pd.read_sql(query8, connection)

plt.figure(figsize=(8, 5))

plt.bar(df8["Region"], df8["Total_Sales"])

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.grid(axis="y")

for i, value in enumerate(df8["Total_Sales"]):
    plt.text(i, value, str(value), ha="center")

plt.tight_layout()

plt.savefig("region_sales.png", dpi=300)

plt.show()
```
