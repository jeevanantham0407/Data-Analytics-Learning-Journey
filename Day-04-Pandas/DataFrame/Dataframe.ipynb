```python
import pandas as pd


data = {
    "Name": ["Arun", "Priya", "Rahul", "Divya"],
    "Age": [21, 22, 20, 21],
    "Marks": [85, 90, 78, 92],
    "Department": ["CSE", "IT", "ECE", "CSE"]
}

df = pd.DataFrame(data)

print("STUDENT DATA:")
print(df)

print("\nSHAPE:")
print(df.shape)

print("\nSIZE:")
print(df.size)

print("\nCOLUMNS:")
print(df.columns)

print("\nINDEX:")
print(df.index)

print("\nDATA TYPES:")
print(df.dtypes)

print("\nFIRST 2 ROWS:")
print(df.head(2))

print("\nLAST 2 ROWS:")
print(df.tail(2))

print("\nSTATISTICAL SUMMARY:")
print(df.describe())

print("\nINFORMATION:")
print(df.info())
```
