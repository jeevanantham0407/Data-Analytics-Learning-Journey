Step 1: Read the Dataset

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\jeeva\OneDrive\Desktop\Data_Analytics_Learning_Journey\Day-04-Pandas\Data_Cleaning\Duplicate Values\Dataset\dirty_duplicates_data.csv")

print("Dirty Dataset:")
print(df)
```

Step 2: Find Duplicate Rows

```python
print("Duplicate Rows:")
print(df.duplicated())
```

Step 3: Display Only Duplicate Rows

```python
print(df[df.duplicated()])
```

Step 4: Count Duplicate Rows

```python
print("Count of duplicate rows:")
print(df.duplicated().sum())
```

Step 5: Remove Duplicate Rows

```python
df_clean = df.drop_duplicates()

print("Cleaned Dataset:")
print(df_clean)

#CHECKING

print("\nOriginal rows:", df.shape[0])
print("\nRows after removing duplicates:", df_clean.shape[0])
```

Step 6: Check Duplicates After Cleaning

```python
print("Dataset after cleaning the duplicates:")
print(df_clean.duplicated().sum())
```

Step 7 — Duplicates Based on a Specific Column

```python
print("Duplicate orderIDs:")
print(df.duplicated(subset=["OrderID"]))

print("\nDuplicate rows:")
print(df[df.duplicated(subset=["OrderID"])])
```

Remove Duplicates Based on OrderID

```python
df_clean = df.drop_duplicates(subset=["OrderID"])

print(df_clean)
```
