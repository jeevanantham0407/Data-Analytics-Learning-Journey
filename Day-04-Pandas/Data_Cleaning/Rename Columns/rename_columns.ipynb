Step 1: Read the Dataset and View Column Names

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\jeeva\OneDrive\Desktop\Data_Analytics_Learning_Journey\Day-04-Pandas\Data_Cleaning\Rename Columns\Dataset\dirty_column_names.csv")

pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)

print("Dataset:")
print(df.to_string())

print("\nColumn Names:")
print(df.columns)
```

Step 2: Rename One Column

```python
df_renamed = df.rename(columns={
    "REGION": "Region"
})

print("Old Column names:")
print(df.columns)

print("\nRenamed Colum Name:")
print(df_renamed.columns)
```

Step 3: Rename Multiple Columns

```python
df_Rename_multiple_columns = df.rename(columns={
    "Order ID": "OrderID",
    "Customer Name": "CustomerName",
    "customer_age": "Age",
    "Product Name": "Product",
    "Product_Category": "Category",
    "REGION": "Region",
    "Total Sales": "Sales"
})

print("Original columns:")
print(df.columns)

print("\nColumns after renamed:")
print(df_Rename_multiple_columns.columns)
```

Step 4: Standardize Column Names Automatically

```python
standardized_df = df.columns.str.lower()

print(standardized_df)
```

Step 5: Replace Spaces with Underscores

```python
df_underscore = df.columns.str.replace(" ", "_")
print(df_underscore)
```

Step 6: Complete Automatic Standardization

```python
cleaned_df = df.columns.str.lower().str.replace(" ","_")

print(cleaned_df)
```
