Load and Inspect the Dataset

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\jeeva\OneDrive\Desktop\Data_Analytics_Learning_Journey\Day-04-Pandas\Mini project-Sales-Data-Cleaning-Analysis\Dataset\dirty_sales_project.csv")

print("Original Dataset:")
print(df.to_string())

print("\nDataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

Handle Missing Values



```python
print("Missing Values:")
print(df.isnull().sum())
```

Handle Categorical Columns


```python

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Sales"] = df["Sales"].fillna(df["Sales"].median())

df["Product"] = df["Product"].fillna(df["Product"].mode()[0])
df["Category"] = df["Category"].fillna(df["Category"].mode()[0])
df["Region"] = df["Region"].fillna(df["Region"].mode()[0])

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

Remove Duplicate Rows


```python
print("Duplicate Rows Before Cleaning:")
print(df.duplicated().sum())

print("\nDuplicate Records:")
print(df[df.duplicated()])

df = df.drop_duplicates()

print("\nDuplicate Rows After Cleaning:")
print(df.duplicated().sum())
```

Rename and Standardize Column Names


```python
print("Column Names Before:")
print(df.columns)

df.columns = df.columns.str.lower().str.replace(" ", "_")

print("\nColumn Names After:")
print(df.columns)
```

Change Data Types


```python
print("Data Types Before:")
print(df.dtypes)

df["order_date"] = pd.to_datetime(df["order_date"])

df["age"] = df["age"].round().astype(int)

print("\nData Types After:")
print(df.dtypes)
```

Data Analysis Using GroupBy

```python
# Total Sales by Region
print("Total Sales by Region:")
print(df.groupby("region")["sales"].sum())

# Total Sales by Product
print("\nTotal Sales by Product:")
print(df.groupby("product")["sales"].sum())

# Total Sales by Category
print("\nTotal Sales by Category:")
print(df.groupby("category")["sales"].sum())

# Total Sales by Salesperson
print("\nTotal Sales by Salesperson:")
print(df.groupby("salesperson")["sales"].sum())

# Region Summary
print("\nRegion Summary:")
print(
    df.groupby("region")["sales"].agg(
        ["sum", "mean", "min", "max", "count"]
    )
)
```

Pivot Table Reports

```python
# Sales by Region and Category
region_category = pd.pivot_table(
    df,
    values="sales",
    index="region",
    columns="category",
    aggfunc="sum",
    fill_value=0
)

print("Sales by Region and Category:")
print(region_category)


# Sales by Region and Product
region_product = pd.pivot_table(
    df,
    values="sales",
    index="region",
    columns="product",
    aggfunc="sum",
    fill_value=0
)

print("\nSales by Region and Product:")
print(region_product)


# Salesperson Performance
salesperson_report = df.groupby("salesperson")["sales"].agg(
    Total_Sales="sum",
    Average_Sales="mean",
    Sales_Count="count"
)

print("\nSalesperson Performance:")
print(salesperson_report)


print("\nSalesperson Performance:")
print(salesperson_report)
```

Final Business Insights

```python
# Total Sales
total_sales = df["sales"].sum()

# Best Region
region_sales = df.groupby("region")["sales"].sum()
best_region = region_sales.idxmax()

# Best Product
product_sales = df.groupby("product")["sales"].sum()
best_product = product_sales.idxmax()

# Best Category
category_sales = df.groupby("category")["sales"].sum()
best_category = category_sales.idxmax()

# Best Salesperson
salesperson_sales = df.groupby("salesperson")["sales"].sum()
best_salesperson = salesperson_sales.idxmax()

print("Total Sales:", total_sales)
print("Best Region:", best_region)
print("Best Product:", best_product)
print("Best Category:", best_category)
print("Best Salesperson:", best_salesperson)
```
