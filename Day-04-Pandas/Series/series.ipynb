```python

import pandas as pd

# Create a Pandas Series
marks = pd.Series([85, 90, 78, 92, 88])

# Display the Series
print("Marks:")
print(marks)

# Access values
print("\nFirst mark:")
print(marks.iloc[0])

# Basic properties
print("\nNumber of values:")
print(marks.size)

print("\nData type:")
print(marks.dtype)

# Statistical operations
print("\nTotal:")
print(marks.sum())

print("\nAverage:")
print(marks.mean())

print("\nHighest mark:")
print(marks.max())

print("\nLowest mark:")
print(marks.min())

# Statistical summary
print("\nStatistical Summary:")
print(marks.describe())
```
