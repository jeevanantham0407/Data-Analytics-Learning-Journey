```python
import pandas as pd

df=pd.read_csv(r"C:\Users\jeeva\OneDrive\Desktop\Data_Analytics_Learning_Journey\Day-04-Pandas\Reading_CSV_File\Dataset\student.csv")

print("ORIGINAL DATA:")
print(df)

# Sort marks in ascending order
print("\nMARKS - ASCENDING:")
print(df.sort_values("Marks"))

# Sort marks in descending order
print("\nMARKS - DESCENDING:")
print(df.sort_values("Marks", ascending=False))

# Sort names alphabetically
print("\nNAMES - A TO Z:")
print(df.sort_values("Name"))

# Sort names in reverse alphabetical order
print("\nNAMES - Z TO A:")
print(df.sort_values("Name", ascending=False))

# Sort by age
print("\nAGE - ASCENDING:")
print(df.sort_values("Age"))

# Sort by department and marks
print("\nDEPARTMENT AND MARKS:")
print(
    df.sort_values(
        ["Department", "Marks"],
        ascending=[True, False]
    )
)

# Filter students above 80 and sort by marks
print("\nMARKS ABOVE 80 - HIGHEST FIRST:")
result = df[df["Marks"] > 80].sort_values(
    "Marks",
    ascending=False
)
print(result)

# Select Name and Marks, filter and sort
print("\nNAME AND MARKS - ABOVE 80:")
result = df.loc[
    df["Marks"] > 80,
    ["Name", "Marks"]
].sort_values(
    "Marks",
    ascending=False
)
print(result)

# Reset the index
print("\nSORTED DATA WITH RESET INDEX:")
result = df.sort_values(
    "Marks",
    ascending=False
).reset_index(drop=True)

print(result)
```
