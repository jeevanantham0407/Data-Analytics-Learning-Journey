1.Electricity Bill Calculator

```python
units = 250

if units <= 100:
    bill = units * 2
elif units <= 200:
    bill = (100 * 2) + ((units - 100) * 3)
else:
    bill = (100 * 2) + (100 * 3) + ((units - 200) * 5)

print("Electricity Bill: ₹", bill)
```

2.Income Tax Calculator

```python
salary = 850000

if salary <= 500000:
    tax = 0
elif salary <= 1000000:
    tax = salary * 0.10
else:
    tax = salary * 0.20

print("Income Tax: ₹", tax)
```

3.ATM Withdrawal System

```python
balance = 10000
withdraw = 3500

if withdraw <= balance:
    if withdraw % 100 == 0:
        if balance - withdraw >= 500:
            print("Transaction Successful")
            print("Remaining Balance:", balance - withdraw)
        else:
            print("Minimum balance of ₹500 must remain.")
    else:
        print("Amount should be in multiples of ₹100.")
else:
    print("Insufficient Balance.")
```

4.Login System with Account Lock

```python
correct_username = "admin"
correct_password = "python123"

attempts = 0

while attempts < 3:
    username = input("Enter Username: ")
    password = input("Enter Password: ")

    if username == correct_username and password == correct_password:
        print("Login Successful")
        break
    else:
        attempts += 1
        print("Invalid Credentials")

if attempts == 3:
    print("Account Locked")
```

5.Student Scholarship Eligibility
