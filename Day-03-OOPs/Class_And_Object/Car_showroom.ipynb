```python
class Car:
    def __init__(self,brand,model,price):
        self.brand = brand
        self.model = model
        self.price = price

    def resale_value(self):
        value = self.price * 0.80
        return value

    def display(self):
        print("Brand:",self.brand)
        print("Model:",self.model)
        print("Price:",self.price)
        print("Resale Value:", self.resale_value())

car1= Car("Hyndai","Creta",1500000)
car1.display()
        
```
