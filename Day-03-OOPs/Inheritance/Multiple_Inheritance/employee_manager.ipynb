```python

class Employee:

    def __init__(self, emp_id, name):
        self.emp_id = emp_id
        self.name = name

    def display_employee(self):
        print("Employee ID :", self.emp_id)
        print("Employee Name :", self.name)


class Salary:

    def __init__(self, monthly_salary):
        self.monthly_salary = monthly_salary

    def calculate_salary(self):
        annual_salary = self.monthly_salary * 12
        print("Monthly Salary :", self.monthly_salary)
        print("Annual Salary :", annual_salary)



class Manager(Employee, Salary):

    def __init__(self, emp_id, name, monthly_salary, department):

        Employee.__init__(self, emp_id, name)
        Salary.__init__(self, monthly_salary)

        self.department = department

    def display_manager(self):
        self.display_employee()
        self.calculate_salary()
        print("Department :", self.department)



manager = Manager(101, "Jeevanantham", 50000, "Data Analytics")

print("------ Manager Details ------")
manager.display_manager()
```
