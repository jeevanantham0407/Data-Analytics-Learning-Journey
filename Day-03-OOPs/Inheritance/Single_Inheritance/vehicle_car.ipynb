```python

class Vehicle:

    def __init__(self, brand, year):
        self.brand = brand
        self.year = year

    def start(self):
        print(f"{self.brand} vehicle started.")

    def stop(self):
        print(f"{self.brand} vehicle stopped.")

    def display_vehicle(self):
        print("Brand :", self.brand)
        print("Year  :", self.year)



class Car(Vehicle):

    def __init__(self, brand, year, model, price):
        super().__init__(brand, year)
        self.model = model
        self.price = price

    def display_car(self):
        self.display_vehicle()
        print("Model :", self.model)
        print("Price :", self.price)

    def mileage(self, distance, fuel):
        mileage = distance / fuel
        print("Mileage :", mileage, "km/l")



car1 = Car("Hyundai", 2024, "Creta", 1800000)

print("------ Car Details ------")
car1.display_car()

print("\n------ Car Operations ------")
car1.start()
car1.mileage(450, 30)
car1.stop()
```
