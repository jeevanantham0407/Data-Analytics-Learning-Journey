```python
class Payment:

    def pay(self, amount):
        print("Payment Processing")


class CreditCard(Payment):

    def pay(self, amount):
        print(f"Paid ₹{amount} using Credit Card")


class UPI(Payment):

    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI")


class Cash(Payment):

    def pay(self, amount):
        print(f"Paid ₹{amount} using Cash")


card = CreditCard()
upi = UPI()
cash = Cash()

card.pay(500)
upi.pay(1200)
cash.pay(300)
```
