```python
from abc import ABC, abstractmethod


class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):

    def start(self):
        print("Car started using a key.")


class Bike(Vehicle):

    def start(self):
        print("Bike started using a self-start button.")


car = Car()
bike = Bike()

car.start()
bike.start()
```
