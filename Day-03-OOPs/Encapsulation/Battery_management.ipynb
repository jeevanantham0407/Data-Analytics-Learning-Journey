```python
class Mobile:

    def __init__(self, brand):
        self.brand = brand
        self.__battery = 50

    def charge(self, percentage):
        self.__battery += percentage

        if self.__battery > 100:
            self.__battery = 100

        print("Battery:", self.__battery, "%")


phone = Mobile("Samsung")

phone.charge(30)
phone.charge(40)
```
