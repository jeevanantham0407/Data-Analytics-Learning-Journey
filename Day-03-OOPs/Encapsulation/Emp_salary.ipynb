```python
class Employee:

    def __init__(self, name):
        self.name = name
        self.__salary = 0

    def set_salary(self, salary):
        if salary > 0:
            self.__salary = salary
        else:
            print("Salary must be positive")

    def display(self):
        print("Employee:", self.name)
        print("Salary:", self.__salary)


emp = Employee("Arun")

emp.set_salary(45000)
emp.display()
```
