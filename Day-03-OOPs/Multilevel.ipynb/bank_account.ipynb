```python

class Account:

    def __init__(self, account_number, account_holder, balance):
        self.account_number = account_number
        self.account_holder = account_holder
        self.balance = balance

    def display_account(self):
        print("Account Number :", self.account_number)
        print("Account Holder :", self.account_holder)
        print("Balance        :", self.balance)



class SavingsAccount(Account):

    def __init__(self, account_number, account_holder, balance, interest_rate):
        super().__init__(account_number, account_holder, balance)
        self.interest_rate = interest_rate

    def calculate_interest(self):
        interest = (self.balance * self.interest_rate) / 100
        print("Interest Rate  :", self.interest_rate, "%")
        print("Interest       :", interest)



class PremiumAccount(SavingsAccount):

    def __init__(self, account_number, account_holder, balance, interest_rate, reward_points):
        super().__init__(account_number, account_holder, balance, interest_rate)
        self.reward_points = reward_points

    def display_premium(self):
        self.display_account()
        self.calculate_interest()
        print("Reward Points  :", self.reward_points)



customer = PremiumAccount(
    1001,
    "Jeevanantham",
    50000,
    6.5,
    1200
)

print("------ Premium Account Details ------")
customer.display_premium()
```
