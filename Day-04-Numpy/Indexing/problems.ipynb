1.Access the First Element

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr[0])
```

2.Access the Last Element

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr[-1])
```

3.Access a Row from a 2D Array

```python
import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(arr[1])
```

4.Access a column From a 2D Array

```python
import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(arr[:, 1])
```

5.Access a Specific Element

```python
import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(arr[1, 2])
```
