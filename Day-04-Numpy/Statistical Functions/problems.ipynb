```python
import numpy as np

marks = np.array([70, 80, 90, 60, 100])

print(np.mean(marks))
```

```python
import numpy as np

marks = np.array([70, 90, 60, 100, 80])

print(np.median(marks))
```

```python
import numpy as np

arr = np.array([10, 20, 20, 30, 20, 40, 10])

values, counts = np.unique(arr, return_counts=True)

mode = values[np.argmax(counts)]

print("Mode:", mode)
```

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(np.var(arr))
```

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(np.std(arr))
```

```python
import numpy as np

marks = np.array([50, 60, 70, 80, 90, 100])

print(np.percentile(marks, 25))
```
