Dot Product using np.dot()

```python
import numpy as np

a=np.array([2,4,5])
b=np.array([6,7,9])

result=np.dot(a,b)

print(result)
```

Matrix Multiplication using np.matmul()

```python
import numpy as np

a = np.array([
    [1, 2],
    [3, 4]
])

b = np.array([
    [5, 6],
    [7, 8]
])

result = np.matmul(a, b)

print(result)
```

Matrix Transpose

```python
import numpy as np

arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

result = np.transpose(arr)

print(result)
```

Matrix Inverse using np.linalg.inv()

```python
import numpy as np

arr = np.array([
    [1, 2],
    [3, 4]
])

result = np.linalg.inv(arr)

print(result)
```

Determinant using np.linalg.det()

```python
import numpy as np

arr = np.array([
    [1, 2],
    [3, 4]
])

result = np.linalg.det(arr)

print(result)
```

Eigenvalues using np.linalg.eig()

```python
import numpy as np

arr = np.array([
    [2, 0],
    [0, 3]
])

eigenvalues, eigenvectors = np.linalg.eig(arr)

print("Eigenvalues:", eigenvalues)
print("Eigenvectors:")
print(eigenvectors)
```

Solve Linear Equations using np.linalg.solve()

```python
import numpy as np

A = np.array([
    [1, 1],
    [2, 3]
])

B = np.array([5, 13])

result = np.linalg.solve(A, B)

print(result)
```

Matrix Rank

```python
import numpy as np

arr = np.array([
    [1, 2],
    [2, 4]
])

result = np.linalg.matrix_rank(arr)

print(result)
```
