Find the Shape



```python
import numpy as np

arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(arr.shape)
```

Find the Size



```python
import numpy as np

arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(arr.size)
```

Find the Item Size



```python
import numpy as np

arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(arr.itemsize)
```

Find Total Memory (nbytes)



```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr.nbytes)
```

Display All Array Properties



```python
import numpy as np

arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Array:")
print(arr)

print("Shape      :", arr.shape)
print("Dimensions :", arr.ndim)
print("Size       :", arr.size)
print("Data Type  :", arr.dtype)
print("Item Size  :", arr.itemsize)
print("Total Bytes:", arr.nbytes)
```
