1.Addition

```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a + b)
```

2.Subtraction

```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a - b)
```

3.Multiplication

```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a * b)
```

4.Division

```python
import numpy as np

a = np.array([20, 30, 40])
b = np.array([2, 3, 4])

print(a / b)
```

5.Modulus

```python
import numpy as np

a = np.array([20, 30, 40])
b = np.array([2, 3, 4])

print(a % b)
```

6.Power

```python
import numpy as np

arr = np.array([2, 3, 4])

print(arr ** 2)
```

7.Square Root

```python
import numpy as np

arr = np.array([4, 9, 16, 25])

print(np.sqrt(arr))
```

8.Sum

```python
import numpy as np

arr = np.array([10, 20, 30, 40])

print(np.sum(arr))
```

9.Mean

```python
import numpy as np

arr = np.array([10, 20, 30, 40])

print(np.mean(arr))
```

10.Standard Deviation

```python
import numpy as np

arr = np.array([10, 20, 30, 40])

print(np.std(arr))
```
