Filter Values Greater Than a Number

```python
import numpy as np

arr = np.array([10, 25, 50, 75, 90])

result = arr[arr > 50]

print(result)
```

Filter Even Numbers

```python
import numpy as np

arr = np.array([10, 15, 20, 25, 30, 35])

result = arr[arr % 2 == 0]

print(result)
```

Filter Odd Numbers

```python
import numpy as np

arr = np.array([10, 15, 20, 25, 30, 35])

result = arr[arr % 2 != 0]

print(result)
```

Use np.where()

```python
import numpy as np

arr = np.array([10, 50, 70, 20, 90])

result = np.where(arr > 50,arr,0)

print(result)
```

Replace Values using Boolean Indexing

```python
import numpy as np

marks = np.array([85, 42, 30, 75, 25, 90])

marks[marks < 35] = 0
print(marks)
```
