Save Array using np.save()

```python
import numpy as np

arr = np.array([1,2,3,4,5,6])

np.save("numbers.npy", arr)

print("Array Saved Successfully")
```

Load Array using np.load()

```python
import numpy as np

new_arr = np.load("numbers.npy")

print(new_arr)
```

Save as Text using np.savetxt()

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

np.savetxt("numbers.txt", arr)

print("File saved")
```

Load Text using np.loadtxt()

```python
import numpy as np

arr = np.loadtxt("numbers.txt")

print(arr)
```

Save Multiple Arrays using np.savez()

```python
import numpy as np

names = np.array(["Arun", "Priya", "Rahul"])

marks = np.array([85, 92, 78])

np.savez("students.npz", names=names, marks=marks)

print("Arrays saved")

data = np.load("students.npz")

print(data["names"])
print(data["marks"])
```
