Simple Chart


```python
import matplotlib.pyplot as plt

x= [1,2,3,4,5]
y= [10, 20, 15, 25, 30]

plt.plot(x, y)
plt.show()
```

Add a Title


```python
plt.plot(x, y)

plt.title("Simple Chart")

plt.show()
```

Add X-axis and Y-axis Labels


```python
plt.plot(x, y)

plt.title("Simple Chart")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```

Add Markers


```python
plt.plot(x, y, marker="s")

plt.title("Simple Chart")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```

Add a Grid


```python
plt.plot(x, y, marker="o")

plt.title("Simple Chart")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.grid()

plt.show()
```

Change Figure Size


```python
plt.figure(figsize=(8, 5))

plt.plot(x, y, marker="o")

plt.title("Simple Chart")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.grid()

plt.show()
```
