Basic Line Chart


```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20000, 25000, 22000, 30000, 35000]

plt.plot(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
```

Line Chart with Markers


```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20000, 25000, 22000, 30000, 35000]

plt.plot(months, sales, marker= "o")

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
```

Change Line Style

```python
plt.plot(months, sales, linestyle="--")
```

Complete Line Chart


```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20000, 25000, 22000, 30000, 35000]

plt.figure(figsize=(8, 5))

plt.plot(
    months,
    sales,
    marker="o",
    linestyle="--",
    label="Sales"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.grid()
plt.legend()

plt.show()
```

Practice Task

Create a line chart using:

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"] <br>
revenue = [15000, 18000, 22000, 20000, 25000, 30000]

Requirements:

Figure size (8, 5) <br>
Title: Monthly Revenue Trend <br>
X-axis: Months <br>
Y-axis: Revenue <br>
Add circle markers <br>
Use dashed line style <br>
Add grid <br>
Add legend

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [15000, 18000, 22000, 20000, 25000, 30000]

plt.figure(figsize= (8, 5))

plt.plot(
    months, 
    revenue,
    marker= "o", 
    linestyle= "--", 
    label= "Revenue"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Months")
plt.ylabel("Revenue")

plt.grid()
plt.legend()

plt.show()
```
