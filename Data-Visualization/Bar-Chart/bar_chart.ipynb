Basic Bar Chart


```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mouse", "Keyboard", "Monitor"]
sales = [50000, 15000, 20000, 35000]

plt.bar(products, sales)

plt.title("Sales by Product")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```

Add Grid

```python
plt.bar(products, sales)

plt.title("Sales by Product")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.grid(axis="y")

plt.show()
```

Add Values on Top of Bars

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mouse", "Keyboard", "Monitor"]
sales = [50000, 15000, 20000, 35000]

plt.bar(products, sales)

plt.title("Sales by Product")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.grid(axis="y")

for i, value in enumerate(sales):
    plt.text(i, value, str(value), ha= "center")

plt.show()
```

Horizontal Bar Chart


```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mouse", "Keyboard", "Monitor"]
sales = [50000, 15000, 20000, 35000]

plt.barh(products, sales)

plt.title("Sales by Product")
plt.xlabel("Sales")
plt.ylabel("Products")

plt.show()
```

Practice Task

Create a bar chart using:

products = ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer"] <br>
sales = [50000, 15000, 20000, 35000, 25000]

Requirements:

Figure size (8, 5) <br>
Title: Product Sales <br>
X-axis: Products <br>
Y-axis: Sales <br>
Add grid on Y-axi <br>
Add the sales values on top of each bar

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer"]
sales = [50000, 15000, 20000, 35000, 25000]

plt.figure(figsize= (8, 5))

plt.bar(products, sales)

plt.title("Product Sales")
plt.xlabel("products")
plt.ylabel("sales")

plt.grid(axis= "y")

for i, value in enumerate(sales):
    plt.text(i, value, str(value), ha= "center")

plt.show()
```
