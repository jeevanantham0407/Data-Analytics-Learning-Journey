```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [10000, 15000, 18000, 16000, 22000, 25000]

products = ["Laptop", "Mouse", "Keyboard", "Monitor"]
product_sales = [50000, 15000, 20000, 35000]

plt.figure(figsize=(10, 8))

# First chart
plt.subplot(2, 1, 1)

plt.plot(months, sales, marker="o", linestyle= "--")

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.grid()

# Second chart
plt.subplot(2, 1, 2)

plt.bar(products, product_sales)

plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.grid(axis="y")

plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [15000, 18000, 22000, 20000, 25000, 30000]

products = ["Laptop", "Mouse", "Keyboard", "Monitor"]
sales = [50000, 15000, 20000, 35000]

plt.figure(figsize=(10, 8))

# Chart 1 – Line Chart
plt.subplot(2, 1, 1)

plt.plot(months, revenue, marker="o")

plt.title("Monthly Revenue")
plt.xlabel("Months")
plt.ylabel("Revenue")

plt.grid()

# Chart 2 – Bar Chart
plt.subplot(2, 1, 2)

plt.bar(products, sales)

plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.grid(axis="y")

plt.tight_layout()

plt.show()
```
