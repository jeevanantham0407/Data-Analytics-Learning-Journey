```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

product_a = [10000, 12000, 15000, 14000, 18000, 20000]
product_b = [8000, 10000, 11000, 13000, 15000, 17000]

plt.figure(figsize=(8, 5))

plt.plot(months, product_a, marker="o", label="Product A")
plt.plot(months, product_b, marker="o", label="Product B")

plt.title("Product Sales Comparison")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.grid()
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

laptop_sales = [50000, 55000, 60000, 58000, 65000, 70000]
mobile_sales = [40000, 45000, 48000, 52000, 55000, 60000]

plt.figure(figsize=(8, 5))

plt.plot(months, laptop_sales, marker="o", label="Laptop")
plt.plot(months, mobile_sales, marker="o", label="Mobile")

plt.title("Laptop vs Mobile Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.grid()
plt.legend()

plt.show()
```
