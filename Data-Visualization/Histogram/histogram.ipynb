```python
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75,
         78, 80, 82, 85, 88, 90, 92, 95]

plt.figure(figsize=(8, 5))

plt.hist(marks, bins=5)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.grid(axis="y")

plt.show()
```

Use this dataset:

marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75,
         78, 80, 82, 85, 88, 90, 92, 95]

Requirements

Create a histogram with:

Figure size: (8, 5) <br>
Title: Student Marks Distribution <br>
X-axis: Marks <br>
Y-axis: Number of Students <br>
bins=5 <br>
Grid on Y-axis <br>
Display the chart

```python
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75,
         78, 80, 82, 85, 88, 90, 92, 95]

plt.figure(figsize= (8, 5))

plt.hist(marks, bins=5)

plt.title("Student Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number Of Students")

plt.grid(axis= "y")

plt.show()
```
