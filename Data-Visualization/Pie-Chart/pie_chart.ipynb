Basic Pie Chart


```python
import matplotlib.pyplot as plt

categories = ["Electronics", "Accessories", "Furniture"]
sales = [50000, 30000, 20000]

plt.pie(sales, labels=categories)

plt.title("Sales by Category")

plt.show()
```

Show Percentages


```python
import matplotlib.pyplot as plt

categories = ["Electronics", "Accessories", "Furniture"]
sales = [50000, 30000, 20000]

plt.pie(
    sales,
    labels=categories,
    autopct="%1.1f%%"
)

plt.title("Sales by Category")

plt.show()
```

Add a Figure Size


```python
plt.figure(figsize=(6, 6))

plt.pie(
    sales,
    labels=categories,
    autopct="%1.1f%%"
)

plt.title("Sales by Category")

plt.show()
```

Explode a Slice


```python
explode = [0.1, 0, 0]

plt.pie(
    sales,
    labels=categories,
    autopct="%1.1f%%",
    explode=explode
)
```

Practice Task

Create a pie chart using:

categories = ["Electronics", "Clothing", "Food", "Books"] <br>
sales = [40000, 25000, 20000, 15000]

Requirements:

Figure size (8, 5) <br>
Title: Sales Distribution by Category <br>
Display category labels <br>
Display percentages <br>
Explode the Electronics slice <br>
Display the chart

```python
import matplotlib.pyplot as plt

categories = ["Electronics", "Clothing", "Food", "Books"]
sales = [40000, 25000, 20000, 15000]

plt.figure(figsize= (8, 5))

explode= [0.1, 0, 0, 0]

plt.pie(
    sales, 
    labels= categories, 
    autopct= "%1.1f%%",
    explode=explode
)

plt.title("Sales Distribution by Category")

plt.show()


```
